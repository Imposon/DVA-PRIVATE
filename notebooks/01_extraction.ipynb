# 01 Extraction

Load the raw **Retail Store Sales** dataset, inspect its structure, and confirm that it is suitable for the project.

**Dataset:** `data/raw/retail_store_sales.csv`  
**Source:** Kaggle – Retail Store Sales  
**Goal:** Verify row/column count, column types, missing values, and key variable distributions.

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().resolve().name == 'notebooks' else Path.cwd().resolve()
```

```python
RAW_PATH = PROJECT_ROOT / 'data/raw/retail_store_sales.csv'
df = pd.read_csv(RAW_PATH)
df.head()
```

```python
print('Rows   :', len(df))
print('Columns:', len(df.columns))
print()
df.info()
```

## Missing Value Audit

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_report = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_report[missing_report['missing_count'] > 0]
```

## Duplicate Row Check

```python
dup_count = df.duplicated().sum()
print(f'Duplicate rows: {dup_count} ({dup_count / len(df) * 100:.2f}%)')
```

## Discount Applied Distribution

```python
discount_counts = df['Discount Applied'].value_counts(dropna=False)
discount_pct    = df['Discount Applied'].value_counts(normalize=True, dropna=False).mul(100).round(2)

print('Discount Applied distribution:')
print(pd.DataFrame({'count': discount_counts, 'pct': discount_pct}))
```

## Categorical Column Unique Values

```python
cat_cols = df.select_dtypes(include='object').columns.tolist()
for col in cat_cols:
    print(f'{col}: {df[col].nunique()} unique values  →  {df[col].unique()[:8]}')
```
